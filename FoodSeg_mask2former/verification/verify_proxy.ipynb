```python
# correct version
from flask import Flask, request, jsonify
import base64
from io import BytesIO
from PIL import Image
import os
from datetime import datetime
from pyngrok import ngrok

# Add project path
import sys
sys.path.append("/root/FoodSeg_mask2former")

# Import your food analysis functions
from gradio_app.model_inference_classification2 import predict_masks, vit_food_classification_local

app = Flask(__name__)

@app.route("/receive_image", methods=["POST"])
def receive_image():
    data = request.get_json()
    base64_image = data.get("base64_image")
    prompt = data.get("prompt", "unknown")

    if not base64_image:
        return jsonify({"error": "No image provided"}), 400

    try:
        # Step 1: Decode and save image
        image_data = base64.b64decode(base64_image)
        image = Image.open(BytesIO(image_data))

        save_dir = "/root/FoodSeg_mask2former/verification"
        os.makedirs(save_dir, exist_ok=True)
        filename = f"gemini_{datetime.now().strftime('%Y%m%d_%H%M%S')}.png"
        image_path = os.path.join(save_dir, filename)
        image.save(image_path)
        print(f"✅ Image saved at: {image_path}")

        # Step 2: Run food analysis
        os.chdir("/root/FoodSeg_mask2former")  # Ensure correct working directory
        vis_image, detected_labels, pixel_percentages, total_energy, total_protein, total_fat, total_carbohydrate, total_fiber, total_sugar, total_sodium = predict_masks(image_path)
        top_label = vit_food_classification_local(image_path)

        # Step 3: Format results
        result = {
            "vit_prediction": top_label,
            "detected_labels": detected_labels,
            "pixel_percentages": pixel_percentages,
            "total_energy_kcal": round(total_energy, 2),
            "total_protein_g": round(total_protein, 2),
            "total_fat_g": round(total_fat, 2),
            "total_carbohydrate_g": round(total_carbohydrate, 2),
            "total_fiber_g": round(total_fiber, 2),
            "total_sugar_g": round(total_sugar, 2),
            "total_sodium_mg": round(total_sodium, 2)
        }
        print("Successfully get results.")
        print(result)

        return jsonify({"status": "success", "result": result})

    except Exception as e:
        import traceback
        print(traceback.format_exc())
        return jsonify({"error": str(e)}), 500

# Start ngrok tunnel
public_url = ngrok.connect(6006)
print("🌐 Public URL for AutoDL Web A:", public_url)

# Run Flask app
app.run(host="0.0.0.0", port=6006)
```
