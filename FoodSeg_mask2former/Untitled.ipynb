```python
import os
from PIL import Image
from transformers import pipeline
from huggingface_hub import snapshot_download

# ------------------ Configuration ------------------
HF_ENDPOINT = "https://hf-mirror.com"

# ------------------ Download model using mirror ------------------
local_dir = snapshot_download(
    repo_id="Shresthadev403/food-image-classification",
    repo_type="model",
    endpoint=HF_ENDPOINT
)

# ------------------ Load pipeline from local directory ------------------
pipe = pipeline(
    task="image-classification",
    model=local_dir
)

# ------------------ Example inference function ------------------
def vit_food_classification_local(image_path):
    image = Image.open(image_path).convert("RGB")
    results = pipe(image)
    if results:
        return results[0]["label"]
    return "Unknown"
```

```python
result = vit_food_classification_local("/root/FoodSeg_mask2former/burger.jpg")
print(result)
```
