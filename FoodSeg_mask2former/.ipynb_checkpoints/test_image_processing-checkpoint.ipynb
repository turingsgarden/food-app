```python
import os
import random
import shutil

# Source and destination directories
source_root = "/root/.cache/kagglehub/datasets/kmader/food41/versions/5/images"
destination_dir = "/root/FoodSeg_mask2former/test_images"

# Create destination dir if not exist
os.makedirs(destination_dir, exist_ok=True)

# Loop through each subdirectory
for category in os.listdir(source_root):
    category_path = os.path.join(source_root, category)
    if os.path.isdir(category_path):
        image_files = [
            f for f in os.listdir(category_path)
            if f.lower().endswith((".jpg", ".jpeg", ".png"))
        ]
        if image_files:
            selected_image = random.choice(image_files)
            src_file = os.path.join(category_path, selected_image)

            # Rename to avoid collision: "category.jpg"
            dest_file = os.path.join(destination_dir, f"{category}.jpg")
            shutil.copy(src_file, dest_file)

print("✅ Done! One image per category copied.")
```
