```python
import json

# Load the original list-based JSON
with open("batch_results.json", "r") as f:
    data = json.load(f)  # this is a list, not a dict

# Convert list to dict keyed by the image filename
converted = {}
for entry in data:
    image_name = entry["image"]
    converted[image_name] = entry

# Save to new file
with open("batch_results_by_name.json", "w") as f:
    json.dump(converted, f, indent=2)
```
